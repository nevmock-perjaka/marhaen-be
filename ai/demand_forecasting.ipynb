```python
import os
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import tensorflow as tf

import os
```

```python
df = pd.read_csv('/kaggle/input/datasets/bakitacos/indonesia-e-commerce-sales-and-shipping-20232025/all_months_clean.csv', sep=';')
df.head()
```

```python
df = df[df['Status Pesanan'] == 'Selesai'].copy()

df
```

```python
used_feature = ['total_qty', 'product_categories', 'Waktu Pesanan Dibuat']

df = df[used_feature]

df
```

```python
df['Waktu Pesanan Dibuat'] = pd.to_datetime(df['Waktu Pesanan Dibuat'], errors='coerce')

df = df.dropna(subset=['Waktu Pesanan Dibuat'])
df['Date'] = df['Waktu Pesanan Dibuat'].dt.date

df['product_categories'] = df['product_categories'].str.split(',').str[0]
data = df.groupby(['Date', 'product_categories'])['total_qty'].sum().reset_index()

data.rename(columns={'total_qty': 'daily_total_qty'}, inplace=True)
data
```

```python
from sklearn.preprocessing import MinMaxScaler

data = data.sort_values(by=['product_categories', 'Date']).reset_index(drop=True)

data['rolling_mean'] = data.groupby('product_categories')['daily_total_qty'].transform(lambda x: x.rolling(7, min_periods=1).mean())
data['rolling_std'] = data.groupby('product_categories')['daily_total_qty'].transform(lambda x: x.rolling(7, min_periods=1).std().fillna(0))
data['Date'] = pd.to_datetime(data['Date'])

data['day_of_week'] = data['Date'].dt.dayofweek

features_to_use = ['daily_total_qty', 'rolling_mean', 'rolling_std', 'day_of_week']

def make_multivariate_windows(window, horizon):
    encoder_input = window[:-horizon]

    decoder_target = window[-horizon:, 0:1]
    
    start_token = encoder_input[-1:, 0:1] 

    decoder_input = tf.concat([start_token, decoder_target[:-1]], axis=0)

    return (encoder_input, decoder_input), decoder_target

def create_multivariate_dataset(series, window_size, horizon, batch_size):
    ds = tf.data.Dataset.from_tensor_slices(series)
    ds = ds.window(window_size + horizon, shift=1, drop_remainder=True)
    ds = ds.flat_map(lambda w: w.batch(window_size + horizon))
    ds = ds.map(lambda w: make_multivariate_windows(w, horizon))
    return ds.batch(batch_size).prefetch(tf.data.AUTOTUNE)

WINDOW_SIZE = 31
HORIZON = 5
BATCH_SIZE = 32

train_ds_global = None
val_ds_global = None
test_ds_global = None

global_scaler = MinMaxScaler()

train_data_list = []
for category_name, group in data.groupby('product_categories'):
    df_cat = group[features_to_use].values
    n_data = len(df_cat)
    train_split = int(n_data * 0.8)
    train_data_list.append(df_cat[:train_split])

global_scaler.fit(np.vstack(train_data_list))

for category_name, group in data.groupby('product_categories'):
    
    df_cat = group[features_to_use].values
    
    n_data = len(df_cat)
    train_split = int(n_data * 0.8)
    val_split = int(n_data * 0.9)
    
    if train_split < WINDOW_SIZE + HORIZON:
        continue
        
    train_data = df_cat[:train_split]
    val_data = df_cat[train_split:val_split]
    test_data = df_cat[val_split:]
    
    train_scaled = global_scaler.transform(train_data)
    val_scaled = global_scaler.transform(val_data)
    test_scaled = global_scaler.transform(test_data)
    
    cat_train_ds = create_multivariate_dataset(train_scaled, WINDOW_SIZE, HORIZON, BATCH_SIZE)
    cat_val_ds = create_multivariate_dataset(val_scaled, WINDOW_SIZE, HORIZON, BATCH_SIZE)
    cat_test_ds = create_multivariate_dataset(test_scaled, WINDOW_SIZE, HORIZON, BATCH_SIZE)
    
    if train_ds_global is None:
        train_ds_global = cat_train_ds
        val_ds_global = cat_val_ds
        test_ds_global = cat_test_ds
    else:
        train_ds_global = train_ds_global.concatenate(cat_train_ds)
        val_ds_global = val_ds_global.concatenate(cat_val_ds)
        test_ds_global = test_ds_global.concatenate(cat_test_ds)

train_ds_global = train_ds_global.shuffle(buffer_size=1000)
```

```python
class CustomEarlyStopping(tf.keras.callbacks.Callback):
    def __init__(self, monitor='val_loss', patience=5, min_delta=0.0001):
        super().__init__()
        self.monitor = monitor
        self.patience = patience
        self.min_delta = min_delta
        
        self.wait = 0
        self.best_loss = np.inf 
        self.best_weights = None

    def on_train_begin(self, logs=None):
        self.wait = 0
        self.best_loss = np.inf
        self.best_weights = None
        print(f"[{self.__class__.__name__}] Activated. Monitoring '{self.monitor}' with patience {self.patience}.")

    def on_epoch_end(self, epoch, logs=None):
        current_loss = logs.get(self.monitor)
        
        if current_loss is None:
            print(f"Warning: {self.monitor} is missing! Early stopping cannot function.")
            return

        if current_loss < (self.best_loss - self.min_delta):
            self.best_loss = current_loss
            self.wait = 0
            self.best_weights = self.model.get_weights()
        else:
            self.wait += 1
            print(f"\n[Early Stopping Alert] No improvement in {self.monitor} for {self.wait} epoch(s).")
            
            if self.wait >= self.patience:
                print(f"\n[Early Stopping Triggered] Stopping training at Epoch {epoch + 1}!")
                
                self.model.stop_training = True
                
                if self.best_weights is not None:
                    print(f"--> Restoring model weights from the best epoch (Loss: {self.best_loss:.5f}).")
                    self.model.set_weights(self.best_weights)
```

```python
lstm_units = 64
num_features = len(features_to_use)

encoder_input = tf.keras.layers.Input(shape=(WINDOW_SIZE, num_features), name="encoder_input")
encoder_lstm = tf.keras.layers.LSTM(lstm_units, return_sequences=True, return_state=True, name="encoder_lstm")
encoder_outputs, state_h, state_c = encoder_lstm(encoder_input)
encoder_states = [state_h, state_c]
 
decoder_input = tf.keras.layers.Input(shape=(HORIZON, 1), name="decoder_input")
decoder_lstm = tf.keras.layers.LSTM(lstm_units, return_sequences=True, return_state=True, name="decoder_lstm")
decoder_outputs, state_h_dec, state_c_dec = decoder_lstm(decoder_input, initial_state=encoder_states)

decoder_dropout = tf.keras.layers.Dropout(0.2, name="decoder_dropout")(decoder_outputs)

attention_layer = tf.keras.layers.MultiHeadAttention(num_heads=4, key_dim=lstm_units, name="multi_head_attention")
context_vector = attention_layer(query=decoder_dropout, value=encoder_outputs, key=encoder_outputs)
vector_dropout = tf.keras.layers.Dropout(0.2, name="vector_dropout")(context_vector)
 
concat_layer = tf.keras.layers.Concatenate(axis=-1, name="concatenate_layer")
decoder_combined_output = concat_layer([decoder_dropout, vector_dropout])
 
output_layer = tf.keras.layers.TimeDistributed(tf.keras.layers.Dense(1), name="output")
output = output_layer(decoder_combined_output)
 
model_attention = tf.keras.models.Model(inputs=[encoder_input, decoder_input], outputs=output)
model_attention.summary()

model_attention.compile(
    loss="mse",
    metrics=["mae"],
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3)
)

early_stopping = CustomEarlyStopping(
    monitor='val_loss', 
    patience=10,
)

history_attention = model_attention.fit(
    train_ds_global,
    epochs=100,
    validation_data=val_ds_global,
    callbacks=[early_stopping,]
)

test_loss, test_mae = model_attention.evaluate(test_ds_global)
print(f"Test MAE: {test_mae}")
```

```python
import numpy as np
from sklearn.metrics import mean_absolute_error

print("Predicting on test data...")
scaled_predictions = model_attention.predict(test_ds_global)

print("Extracting actual targets...")
scaled_actuals_list = []
for inputs, targets in test_ds_global:
    scaled_actuals_list.append(targets.numpy())

# Stack all the batches together
scaled_actuals = np.concatenate(scaled_actuals_list, axis=0)

flat_predictions = scaled_predictions.flatten()
flat_actuals = scaled_actuals.flatten()

dummy_pred = np.zeros((len(flat_predictions), 4))
dummy_actual = np.zeros((len(flat_actuals), 4))

dummy_pred[:, 0] = flat_predictions
dummy_actual[:, 0] = flat_actuals

real_predictions = global_scaler.inverse_transform(dummy_pred)[:, 0]
real_actuals = global_scaler.inverse_transform(dummy_actual)[:, 0]

real_predictions = np.maximum(real_predictions, 0)

real_mae = mean_absolute_error(real_actuals, real_predictions)

print(f"The actual Mean Absolute Error is: {real_mae:.2f} items")
```
